The `Cabin` column still has a large number of missing values. Since there are too many missing values in this column, it's best to drop it. Dropping the `Cabin` column will remove it from the DataFrame.

```python
df.drop('Cabin', axis=1, inplace=True)
print("Missing values after dropping 'Cabin' column:")
display(df.isnull().sum())
```

```python
# Survival rate by Gender
survival_by_gender = df.groupby('Sex')['Survived'].mean().reset_index()
print("Survival Rate by Gender:")
display(survival_by_gender)
```

```python
# Survival count by Embarked location
survival_count_by_embarked = df.groupby('Embarked')['Survived'].sum().reset_index()
print("\nSurvival Count by Embarked Location:")
display(survival_count_by_embarked)
```

# Task
Calculate the average age of survivors versus non-survivors, find the highest and lowest fare, calculate the overall survival percentage, and summarize the identified patterns affecting survival from the Titanic dataset.

## Calculate Average Age of Survivors vs. Non-Survivors

### Subtask:
Group the DataFrame by the 'Survived' column and calculate the mean 'Age' for both groups (survivors and non-survivors). Display the result.


**Reasoning**:
The subtask requires grouping the DataFrame by the 'Survived' column and calculating the mean of the 'Age' for each group, then printing the result. I will use the `groupby()` and `mean()` methods to achieve this.



```python
average_age_by_survival = df.groupby('Survived')['Age'].mean()
print("Average Age by Survival Status:")
print(average_age_by_survival)
```

## Find Highest and Lowest Fare

### Subtask:
Find the highest and lowest fare from the 'Fare' column of the DataFrame. Display both values.

**Reasoning**:
To find the highest and lowest fare, I will use the `max()` and `min()` methods directly on the 'Fare' column of the DataFrame. This will provide the required values efficiently.



```python
highest_fare = df['Fare'].max()
lowest_fare = df['Fare'].min()

print(f"Highest Fare: {highest_fare}")
print(f"Lowest Fare: {lowest_fare}")
```

## Calculate Overall Survival Percentage

### Subtask:
Calculate the overall survival percentage from the 'Survived' column of the DataFrame. Display the result as a percentage.

**Reasoning**:
To calculate the overall survival percentage, I will compute the mean of the 'Survived' column and multiply by 100, then display the result.



```python
overall_survival_percentage = df['Survived'].mean() * 100
print(f"Overall Survival Percentage: {overall_survival_percentage:.2f}%")
```
